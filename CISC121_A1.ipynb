# 1. CyclicRotation

```python
def solution(A, K):
    # If the array is empty, return it as is
    if not A:
        return A

    # K can be larger than the size of array, so we take modulo
    n = len(A)
    K = K % n

    # Rotate using slicing
    return A[-K:] + A[:-K]
```

```python
# Test Cases
print(solution([3, 8, 9, 7, 6], 3))
print(solution([0, 0, 0], 1))
print(solution([1, 2, 3, 4], 4))
```

# 2. OddOccurrencesInArray

```python
def solution(A):
    # Initialize result to 0
    unpaired = 0

    # XOR all elements in the array
    # Pairs of identical numbers will cancel each other out
    for number in A:
        unpaired ^= number

    return unpaired
```

```python
# Test Case
test_array = [9, 3, 9, 3, 9, 7, 9]
print(f"Unpaired element: {solution(test_array)}")
```

# 3. FrogJmp

```python
def solution(X, Y, D):
    # Calculate the total distance to cover
    distance = Y - X

    # If the frog is already at or past the target, 0 jumps are needed
    if distance <= 0:
        return 0

    # Calculate minimal jumps using integer division ceiling: (distance + D - 1) // D
    return (distance + D - 1) // D
```

```python
# Test Case
print(f"Minimal jumps: {solution(10, 85, 30)}")
```

# 4. PermMissingElem

```python
def solution(A):
    # N is the length of the array
    n = len(A)

    # The array should contain elements from 1 to N + 1
    # Use the sum formula to find the expected sum of 1 to N + 1
    expected_sum = (n + 1) * (n + 2) // 2

    # Calculate the actual sum of elements in A
    actual_sum = sum(A)

    # The difference is the missing element
    return expected_sum - actual_sum
```

```python
# Test Case
test_array = [2, 3, 1, 5]
print(f"Missing element: {solution(test_array)}")
```

# 5. FrogRiverOne

```python
def solution(X, A):
    # Keep track of covered positions
    covered_positions = set()

    # Iterate through array A where index is the time (seconds) and value is the position
    for time, position in enumerate(A):
        # We only care about positions between 1 and X
        if position <= X:
            covered_positions.add(position)

        # If all positions from 1 to X are covered, return the current time
        if len(covered_positions) == X:
            return time

    # If the frog is never able to jump across the river
    return -1
```

```python
# Test Case
X = 5
A = [1, 3, 1, 4, 2, 3, 5, 4]
print(f"Earliest time: {solution(X, A)}")
```
